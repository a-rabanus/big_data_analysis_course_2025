```python
import sqlite3
import pandas as pd
import os

# --- Configuration ---
# 1. DEFINE YOUR PROJECT'S ROOT DIRECTORY
#    This makes the script more portable.
project_root = r"C:\Users\anton\OneDrive\Documents\HSD\sem4\big_data_analysis_course_2025" #<-- IMPORTANT: Make sure this is the correct root folder for your project

# 2. CREATE THE FULL, ABSOLUTE PATH TO THE DATABASE
db_path = os.path.join(project_root, 'images.db')

print(f"Attempting to connect to database at: {db_path}")

# --- Database Connection ---
try:
    conn = sqlite3.connect(db_path)
    print("Connection successful!")
except sqlite3.Error as e:
    print(f"Error connecting to database: {e}")
    # Stop execution if we can't connect
    raise

# --- SQL Queries ---
total_query = "SELECT COUNT(*) FROM image_characteristics"
null_embedding_query = "SELECT COUNT(*) FROM image_characteristics WHERE embedding IS NULL"
null_examples_query = "SELECT * FROM image_characteristics WHERE embedding IS NULL LIMIT 10"

# --- Execute Queries ---
try:
    total_count = pd.read_sql_query(total_query, conn).iloc[0,0]
    null_count = pd.read_sql_query(null_embedding_query, conn).iloc[0,0]
    null_examples_df = pd.read_sql_query(null_examples_query, conn)
    print("Queries executed successfully!")
except Exception as e:
    print(f"An error occurred during query execution: {e}")
    
# --- Close Connection ---
conn.close()

# --- Display Results ---
if 'total_count' in locals():
    print("---")
    print(f"Total rows in 'image_characteristics': {total_count}")
    print(f"Rows with NULL embeddings: {null_count}")

    if total_count > 0:
        percentage = (null_count / total_count) * 100
        print(f"Percentage of NULL embeddings: {percentage:.2f}%")

    print("\n--- Examples of rows with NULL embeddings ---")
    display(null_examples_df)
```
